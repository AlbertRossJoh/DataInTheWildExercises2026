```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

### Load data

```python
# load the data from csv
df = pd.read_csv('us_states_geo.csv')
```

### Basic data exploration

```python
# list all the columns
df.columns
```

```python
# print the shape of the dataframe (#rows, #columns)
df.shape
```

```python
# lists how many entries in are not-null in each column.
# it also list the types associated with each column.
# types are automatically inferred by pandas when the file is loaded, 
# but can be declared explicitly in the read_csv method
df.info()
```

```python
# when addressing a specific field, a pandas Series is returned
# e.g., df.population  OR  df['population']
# Series objects include some simple stats methods, such as max and min
print(df.population.min(), df.population.max(),  df.population.mean())

# series can be passed as iterables to other methods 
print(np.mean(df.population))

# basic statistics for all columns
df.describe()
```

```python
# selects a subset of columns and prints the first N entries
df[['state', 'state_code', 'capital', 'lat_centroid', 'lon_centroid', 'population']].head(6)
```

```python
# number of unique values in a column
len(df.state_code.unique())
```

### Frequency distributions

```python
df.gdp_pc.hist(bins=15)
```

```python
df.uninsured.hist(bins=15)
```

```python
# if you are familiar with matplotlib, you can visualize slightly better plots

# create a figure with subfigures: 1 row and 2 columns
# the axs variable is a list (or more generally a matrix) containing the suboplots
fig, axs = plt.subplots(1,2, figsize=(7,3)) 

df.gdp_pc.hist(bins=15, ax=axs[0]) # plot the histogram into the first subplot (axs[0])
df.uninsured.hist(bins=15, ax=axs[1]) # plot the histogram into the second subplot (axs[1])

# some aesthetic adjustments
for ax in axs:
    ax.grid(False)
axs[0].set_ylabel('frequency')
axs[0].set_xlabel('GDP per capita')
axs[1].set_xlabel('% Uninsured')
plt.tight_layout()

# save the plot on file
plt.savefig('hist.png', dpi=400)
```

```python
# plots a scatter matrix, that contains histograms on the diagonal
# and scatterplots showing the relationship between pairs of variables
# in the other cells
field_selection = ['population','%poor', 'gdp_pc', 'uninsured']
pd.plotting.scatter_matrix(df[field_selection], diagonal='hist')
```
