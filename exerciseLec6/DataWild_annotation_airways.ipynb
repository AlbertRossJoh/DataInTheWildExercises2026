## Data in the Wild - Annotation Exercises - Part 2 Airways

In these exercises you will use some existing crowdsourced data, to practice working with inter-observer agreement.

This dataset contains crowdsourced segmentations measurements of airways in slices of chest CT images. An airway appears like a dark circle or ellipse, with a lighter ring around it. You can see examples of airways and read more about the study [in this paper](https://journals.plos.org/plosone/article?id=10.1371/journal.pone.0249580).

Each annotation *task* is an image, which contains an airway. The annotator has to draw two ellipses: one inside the airway, and one outside the airway. The combination of a specific task and a specific annotator is a *result*, which contains two ellipses.

The original crowdsourced data was stored as ellipse coordinates drawn by the annotators, you can find the images and the raw data here: https://github.com/adriapr/crowdairway

For this exercise, this data has already been cleaned (removing some invalid annotations) and processed to measure how large the ellipses are. The CSV therefore contains:

* task_id - ID of the image
* annotator_id - anonymized ID of the annotator
* result_id - ID of this task/annotator combination
* outer - the area of the outer ellipse
* inner - the area of the inner ellipse

There is also a CSV with expert annotations for these airways with:
* task_id - ID of the image
* subject_id - which patient has this airway
* airway_id - index of the airway in that patient
* generation - the generation of the airway, higher generation airways are smaller
* outer1, outer2 - outer ellipse annotated by experts 1 and 2
* inner1, inner2 - inner ellipse annotated by experts 1 and 2

## Question 2: Combining results per task

We now have a case where not all annotators annotated all images.

If we are interested how the crowd did overall, we want to combine their results into a single measurement per task, and compare it to the experts.

The following code combines the measurements for each task using the *mean* function.

```python

task_list = []

for task_id in df_res['task_id'].unique():
    task_results = df_res.loc[df_res['task_id'] == task_id]

    outer_mean = task_results['outer'].mean()
    inner_mean = task_results['inner'].mean()

    task_dict = {
                'task_id':      task_id,
                'num_combined': len(task_results),
                'outer_mean': outer_mean,
                'inner_mean': inner_mean,
                }
    task_list.append(task_dict)

df_task_combined = pd.DataFrame(task_list)
df_task_combined.head(5)
```
