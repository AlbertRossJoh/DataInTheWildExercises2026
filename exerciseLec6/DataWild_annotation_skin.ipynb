## Other measures

**Note**: the original Kappa score is only for two annotators and categorical variables. For extensions of the Cohen's Kappa you will need to adapt your own function, or use other packages, for example:

* Weighted Cohen's Kappa for ordinal variables: https://datatab.net/tutorial/weighted-cohens-kappa

* Fleiss' Kappa https://en.wikipedia.org/wiki/Fleiss%27_kappa for categorical variables with multiple annotators

* Kendall's tau https://en.wikipedia.org/wiki/Kendall_rank_correlation_coefficient for ordinal variables

Always check the assumptions of a specific measure and why you should / should not apply it to your data! The papers on LearnIT should also be helpful with this.
